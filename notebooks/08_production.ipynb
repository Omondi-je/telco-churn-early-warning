```python
import requests
import json

# Start API first: docker-compose up --build
BASE_URL = "http://localhost:8000"

# Health check
r = requests.get(f"{BASE_URL}/health")
print(r.json())

# Predict
customer = {
    "gender": "Female",
    "SeniorCitizen": 0,
    "Partner": "No",
    "Dependents": "No",
    "tenure": 2,
    "PhoneService": "Yes",
    "MultipleLines": "No",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "No",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "No",
    "StreamingMovies": "No",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 70.0,
    "TotalCharges": 140.0
}

r = requests.post(f"{BASE_URL}/predict", json=customer)
print(json.dumps(r.json(), indent=2))
```

```python
import time

# Batch test
customers = [customer for _ in range(100)]

start = time.time()
for c in customers:
    requests.post(f"{BASE_URL}/predict", json=c)

elapsed = time.time() - start
print(f"100 predictions in {elapsed:.2f}s ({100/elapsed:.1f} req/sec)")
```
